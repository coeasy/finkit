# Formula Engine Demo

AlphaTA includes a TongDaXin-compatible formula engine that lets you write
indicator expressions as strings and evaluate them against OHLCV data.

```python
import numpy as np
import finkit as ta
```

```python
# Generate sample data
np.random.seed(42)
n = 200
close = np.cumsum(np.random.randn(n) * 0.5) + 100.0
high = close + np.abs(np.random.randn(n) * 0.3)
low = close - np.abs(np.random.randn(n) * 0.3)
open_ = close + np.random.randn(n) * 0.1
volume = np.random.uniform(1000, 5000, n)
```

```python
# Evaluate a simple formula
result = ta.formula_eval(
    "MA5:=MA(CLOSE,5); MA10:=MA(CLOSE,10); CROSS(MA5,MA10)",
    open_.tolist(), high.tolist(), low.tolist(),
    close.tolist(), volume.tolist()
)

print("Output variables:", list(result.keys()))
print(f"MA5 last 5: {result.get('MA5', [])[-5:]}")
print(f"Golden crosses: {sum(1 for v in result['__result__'] if v == 1.0)}")
```

```python
# Multi-output formula: RSI + Bollinger Bands
formula = """RSI14:=RSI(CLOSE,14);
UPPER:=MA(CLOSE,20)+2*STD(CLOSE,20);
LOWER:=MA(CLOSE,20)-2*STD(CLOSE,20);
RSI14>70 AND CLOSE>UPPER"""

result = ta.formula_eval(
    formula,
    open_.tolist(), high.tolist(), low.tolist(),
    close.tolist(), volume.tolist()
)
print("Variables:", [k for k in result.keys() if k != '__result__'])
```

```python
# Validate formula syntax without execution
valid = ta.formula_validate("MA(CLOSE, 5)")
invalid = ta.formula_validate("MA(CLOSE,")
print(f"'MA(CLOSE, 5)' valid: {valid}")
print(f"'MA(CLOSE,' valid: {invalid}")
```
