# AlphaTA Quick Start

This notebook demonstrates basic installation and indicator computation with AlphaTA.

```python
import numpy as np
import finkit as ta

# Generate sample price data
np.random.seed(42)
close = np.cumsum(np.random.randn(200)) + 100.0
print(f"Data length: {len(close)}")
print(f"Price range: {close.min():.2f} - {close.max():.2f}")
```

```python
# Compute SMA and EMA
sma_20 = ta.sma(close, timeperiod=20)
ema_12 = ta.ema(close, timeperiod=12)

print(f"SMA(20) last 5: {sma_20[-5:]}")
print(f"EMA(12) last 5: {ema_12[-5:]}")
```

```python
# Compute RSI
rsi_14 = ta.rsi(close, timeperiod=14)
print(f"RSI(14) last value: {rsi_14[-1]:.2f}")
print(f"Overbought bars (RSI > 70): {np.sum(np.array(rsi_14) > 70)}")
print(f"Oversold bars (RSI < 30): {np.sum(np.array(rsi_14) < 30)}")
```

```python
# Compute MACD
macd = ta.macd(close, fastperiod=12, slowperiod=26, signalperiod=9)
print(f"MACD line last: {macd.macd[-1]:.4f}")
print(f"Signal line last: {macd.signal[-1]:.4f}")
print(f"Histogram last: {macd.hist[-1]:.4f}")
```
